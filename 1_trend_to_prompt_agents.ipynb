# 🚀 Proyecto: Trend-to-Prompt Agent (CrewAI)

Este proyecto utiliza un agente inteligente para analizar tendencias de moda y convertirlas en **prompts optimizados** para la API de **Leonardo.ai**.

### Objetivos:
- Identificar estilos visuales clave.
- Traducir conceptos de eventos/tendencias a parámetros técnicos.
- Generar gráficos vectoriales listos para estampado en camisetas.

```python
import os
from crewai import Agent, Task, Crew, LLM

# Configuración de Ollama Local (Llama 3.2 1B)
ollama_llm = LLM(
    model="ollama/llama3.2:1b",
    base_url="http://localhost:11434"
)
```

## 🤖 Configuración del Agente y Tarea: Prompt Engineer

En esta sección definimos el **agente especializado** en ingeniería de prompts para modelos de generación visual (*Leonardo.ai*) y la **tarea específica** que debe ejecutar.

* **Rol del Agente:** Diseñador y experto en prompts de arte digital para camisetas.
* **Objetivo:** Traducir conceptos brutos de tendencias en instrucciones técnicas en inglés (*estilo vectorial, paletas de colores, isolations*).
* **Entregable:** Un prompt optimizado listo para la API/interfaz de Leonardo.ai.

```python
# 1. Agente especializado en Prompts de Leonardo.ai
prompt_engineer = Agent(
    role="Ingeniero de Prompts para Leonardo.ai",
    goal="Convertir ideas de tendencias en camisetas en prompts técnicos altamente detallados para Leonardo.ai.",
    backstory=(
        "Eres un experto en arte digital y diseño de moda urbana. Sabes exactamente qué palabras clave "
        "usar en Leonardo.ai (estilos de iluminación, paletas de colores, gráficos vectoriales, isolations "
        "en fondo blanco) para generar estampados de camisetas que se vendan muy bien."
    ),
    verbose=True,
    llm=ollama_llm
)

# 2. Tarea asignada al Agente
task_generate_prompt = Task(
    description=(
        "Toma la siguiente tendencia de moda/evento: '{tendencia}'. "
        "Crea un prompt súper detallado en INGLÉS listo para copiar y pegar en Leonardo.ai. "
        "Incluye detalles de estilo visual, tipo de ilustración (vector/flat design), colores principales y palabras clave de calidad."
    ),
    expected_output="Un prompt técnico detallado en inglés optimizado para Leonardo.ai, seguido de una breve explicación en español de por qué elegiste esos elementos.",
    agent=prompt_engineer
)
```

## 🚀 Orquestación y Ejecución de la Crew

En este paso ensamblamos la **Crew** uniendo nuestro agente (*Prompt Engineer*) con su tarea asignada y lanzamos el flujo de trabajo pasando un caso de prueba inicial.

* **Modo Verbose:** Activado para visualizar el razonamiento del agente en tiempo real.
* **Entrada de Prueba:** "Festival de música Cyberpunk con neón e ilustraciones de gatos".
* **Resultado Esperado:** Salida técnica estructurada para usar directamente en la API o interfaz de *Leonardo.ai*.

```python
# Instanciar el equipo
crew = Crew(
    agents=[prompt_engineer],
    tasks=[task_generate_prompt],
    verbose=True
)

# Ejecutar con una prueba concreta
resultado = await crew.kickoff_async(inputs={'tendencia': 'Festival de música Cyberpunk con neón e ilustraciones de gatos'})
print("\n================ RESULTADO FINAL ================\n")
print(resultado)
```

## 🕵️‍♂️ Paso 2: Multi-Agent Crew (Trend Researcher + Prompt Engineer)

En este paso escalamos la arquitectura a una **Crew Multi-Agente**:
1. **Trend Researcher Agent:** Analiza nichos, eventos o tendencias de moda urbana y extrae los elementos visuales clave.
2. **Prompt Engineer Agent:** Toma el análisis del primer agente y lo traduce en un prompt técnico para *Leonardo.ai*.

```python
from crewai import Agent, Task, Crew, LLM

# 1. Definir los 2 Agentes
trend_researcher = Agent(
    role="Investigador de Tendencias de Moda e Indumentaria",
    goal="Identificar elementos visuales clave, paletas de colores y temáticas populares en un evento o nicho.",
    backstory=(
        "Eres un analista de tendencias especializado en streetwear y dropshipping de camisetas. "
        "Sabes identificar qué elementos gráficos hacen que un diseño de camiseta se vuelva viral "
        "y tenga alta demanda comercial."
    ),
    verbose=True,
    llm=ollama_llm
)

prompt_engineer = Agent(
    role="Ingeniero de Prompts para Leonardo.ai",
    goal="Convertir análisis de tendencias en prompts técnicos detallados en inglés para Leonardo.ai.",
    backstory=(
        "Eres un experto en arte digital y prompt engineering para Leonardo.ai. "
        "Transformas conceptos de diseño en instrucciones técnicas estructuradas "
        "(vector style, isolated white background, lighting, color palette)."
    ),
    verbose=True,
    llm=ollama_llm
)

# Tareas optimizadas para evitar falsos positivos
task_research = Task(
    description=(
        "Analiza este tema para estampado de ropa: '{evento}'. "
        "Lista 3 elementos visuales: estilo artístico, elementos gráficos principales y colores recomendados."
    ),
    expected_output="Un resumen de 3 puntos con la propuesta de arte para la camiseta.",
    agent=trend_researcher
)

task_generate_prompt = Task(
    description=(
        "Toma la propuesta de arte y redacta un prompt en INGLÉS para Leonardo.ai. "
        "Especifica estilo vectorial, fondo blanco aislado y detalles de ilustración para camisetas."
    ),
    expected_output="Un prompt técnico detallado en inglés optimizado para Leonardo.ai.",
    agent=prompt_engineer
)

# 3. Orquestar la Crew con 2 agentes y 2 tareas
crew_multi_agent = Crew(
    agents=[trend_researcher, prompt_engineer],
    tasks=[task_research, task_generate_prompt],
    verbose=True
)

# 4. Ejecutar el pipeline completo
resultado_final = await crew_multi_agent.kickoff_async(
    inputs={'evento': 'Arte retro de carreras de autos estilo años 80'}
)

print("\n================ RESULTADO FINAL PIPELINE ================\n")
print(resultado_final)
```
